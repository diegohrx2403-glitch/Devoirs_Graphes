### **3. Algorithme**

**Consignes:** Dans ce notebook, vous devez encoder une fonction `parcours_eulerien` qui prend en entrée un graphe simple et non-dirigé, et retourne en sortie un parcours eulérien si celui-ci existe et `None` si un tel parcours n'existe pas. Le graphe donné en entrée sera toujours connexe mais pas forcément eulérien ni même semi-eulérien (NB: un graphe eulérien est un graphe qui admet un parcours eulérien fermé, et qui possède donc exclusivement des noeuds de degré pair. Un graphe semi-eulérien est un graphe qui admet un parcours eulérien, pas forcément fermé, et qui possède donc soit exactement 0 soit exactement 2 noeuds de degré impair).

La réponse attendue doit être représentée par une liste des indices des noeuds du parcours emprunté (qu'il soit ouvert ou fermé). Si le parcours est fermé, le noeud de départ et de fin doit être indiqué deux fois. Pour le graphe 1, une solution possible est :

$$
[0,1,2,3,6,4,7,8,9,5,4,3]
$$

Comme précisé précédemment, s'il s'avère que le graphe ne possède aucun parcours eulérien, la fonction devrait répondre `None`. Attention : votre code ne doit pas modifier le graphe donné en entrée. Si vous le souhaitez, vous pouvez par contre copier le graphe et modifier la copie (la fonction `deepcopy` du module standard `copy` pourrait vous être utile pour cette partie).



**Indice:** Étant donné un graphe (semi-)eulérien, une manière d'implémenter l'algorithme de Hierholzer consiste à parcourir chaque arête deux fois : une fois dans le sens "aller" et une fois dans le sens "retour". Les arêtes sont ajoutées au parcours eulérien (en construction) lorsqu'elles sont parcourues dans le sens "retour".

Plus précisément, en partant d'un nœud susceptible d'être le premier (ou le dernier) d'un parcours eulérien, on parcourt des arêtes dans le sens "aller" jusqu'à atteindre un nœud qui ne possède plus d'arêtes non visitées. On obtient alors une parcours sans arêtes répétées, mais qui ne passe pas nécessairement par toutes les arêtes du graphe : à ce stade, il est possible qu'il reste des parcours fermés non explorés.

Pour trouver ces éventuels parcours fermés, on parcourt alors les arêtes déjà visitées dans le sens inverse, c'est-à-dire dans le sens "retour", jusqu'à atteindre un nœud qui possède encore des arêtes non visitées. Les arêtes parcourues dans le sens "retour" sont ajoutées au parcours eulérien en construction.

Lorsqu'on atteint un tel nœud, on sait qu'un parcours fermé non exploré peut être parcouru à partir de celui-ci. On veut l'ajouter au sein de notre parcours eulérien en construction. On interrompt donc le processus "retour" et on recommence à parcourir des arêtes non encore explorées dans le sens "aller". À nouveau, on continue jusqu'à atteindre un nœud qui ne possède plus d'arêtes non visitées (pour les plus observateurs, il s'agit précisément du nœud sur lequel on s'était arrêté lors du processus "retour").

On a ainsi parcouru un parcours fermé supplémentaire, mais il est possible qu'il reste encore d'autres parcours fermés non explorés. C'est pourquoi on alterne ainsi les parcours "retour" et "aller" jusqu'à ce que toutes les arêtes aient été visitées dans les deux sens.

Ainsi, dans cette implémentation, chaque arête est parcourue deux fois : une fois dans le sens "aller" et une fois dans le sens "retour". Elle n'est ajoutée qu'une seule fois au parcours eulérien en construction, lors du passage "retour".

Une implémentation concrète est possible, par exemple, en maintenant deux piles (ou listes) : l'une contenant les arêtes parcourues dans le sens "aller" (et qui n'ont pas encore été parcourues dans le sens "retour"), et l'autre contenant les arêtes parcourues dans le sens "retour". Cette seconde liste nous donnera le parcours eulérien à la fin de l'algorithme.

En pratique, on maintiendra plutôt deux listes de nœuds, qui représentent les extrémités des arêtes concernées.

Par exemple, sur le graphe 1, une exécution possible est la suivante :
- Etape 0 - vérifier si le graphe est bien (semi-)eulérien, et choisir un noeud de départ (Si le graphe est eulérien, n'importe quel noeud suffit. S'il est semi-eulérien, il faut choisir un des deux noeuds de degré impair. Ici, on choisit le noeud 0).
- Début de l'étape 1 (initialisation)   - liste "aller": [0], liste "retour": []. On place le noeud de départ dans la liste "aller". On commence l'étape 1, qui correspond à un parcours "aller".
- Fin de l'étape 1, début de l'étape 2  - liste "aller": [0,1,2,3,6,4,3], liste "retour": []. Le noeud 3 n'ayant plus d'arêtes non-visitées, on commence un parcours "retour" (l'étape 2).
- Fin de l'étape 2, début de l'étape 3  - liste "aller": [0,1,2,3,6,4], liste "retour": [3]. Lors du parcours "retour", on arrive au sommet 4, qui possède encore une arête inutilisée. On interrompt donc le retour à cet endroit et on reprend un parcours "aller" depuis 4.
- Fin de l'étape 3, début de l'étape 4  - liste "aller": [0,1,2,3,6,4,7,8,9,5,4], liste "retour": [3]. Le noeud 4 n'ayant plus d'arêtes non-visitées, on commence un parcours "retour".
- Fin de l'étape 4, fin de l'algorithme - liste "aller": [], liste "retour": [3,4,5,9,8,7,4,6,3,2,1,0]. Aucun des nœuds rencontrés lors du parcours "retour" ne possède encore d'arête non visitée. Toutes les arêtes du graphe ont donc été parcourues dans les deux sens : on arrête l'algorithme. La liste "retour" donne alors un parcours eulérien du graphe.

### 5. La structure du graphe a-t-elle une influence ?

Réalisez maintenant la même analyse de complexité sur les graphes ci-dessous.

```python
def graphe_cycle(n):
    """
    Génère un graphe en cycle avec n sommets.
    """
    graph = [[] for _ in range(n)]
    for i in range(n):
        graph[i].append((i + 1) % n)
        graph[(i + 1) % n].append(i)
    return graph

def graphe_complet(n):
    """
    Génère un graphe complet avec n sommets, ou n+1 sommets si n est pair (on veut un graphe eulérien).
    """
    if n%2 == 0:
        n += 1
    graph = [[] for _ in range(n)]
    for i in range(n):
        for j in range(n):
            if i != j:
                graph[i].append(j)
    return graph
```
